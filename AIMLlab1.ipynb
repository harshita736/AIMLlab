```python

# Breadth First Search

graph = {
    'A': ['B', 'C'],
    'B': ['D', 'E'],
    'C': ['F'],
    'D': [],
    'E': ['F'],
    'F': []
}

def bfs(graph, start):
    visited = []
    queue = [start]

    while queue:
        node = queue.pop(0)

        if node not in visited:
            print(node, end=" ")
            visited.append(node)

            for neighbour in graph[node]:
                if neighbour not in visited:
                    queue.append(neighbour)

# Main program
print("BFS Traversal:")
bfs(graph, 'A')


# Depth First Search

graph = {
    'A': ['B', 'C'],
    'B': ['D', 'E'],
    'C': ['F'],
    'D': [],
    'E': ['F'],
    'F': []
}

def dfs(graph, start, visited=None):
    if visited is None:
        visited = []

    if start not in visited:
        print(start, end=" ")
        visited.append(start)

        for neighbour in graph[start]:
            dfs(graph, neighbour, visited)

# Main program
print("DFS Traversal:")
dfs(graph, 'A')

```
